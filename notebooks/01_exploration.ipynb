```python
print("Categories in Domain :   ",end=" ")
print(df['Domain'].unique())

print("Categories in Country :   ",end=" ")
print(df['Country'].unique())

print("Categories in City :   ",end=" ")
print(df['City'].unique())

print("Categories in Funding stage :   ",end=" ")
print(df['Funding_Stage'].unique())

print("Categories in Investor Tier :   ",end=" ")
print(df['Investor_Tier'].unique())

print("Categories in AI Adoption Level :   ",end=" ")
print(df['AI_Adoption_Level'].unique())

print("Categories in Acquisition status :   ",end=" ")
print(df['Acquisition_Status'].unique())
```

```python
## define numeric and categorical features
numeric_features=[feature for feature in df.columns if pd.api.types.is_numeric_dtype(df[feature])]
categorical_features=[feature for feature in df.columns if pd.api.types.is_categorical_dtype(df[feature]) 
                      or pd.api.types.is_object_dtype(df[feature])]

print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```
